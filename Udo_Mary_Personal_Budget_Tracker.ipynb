# Personal Budget Tracker

**Why I built this**

Built to track and budget right.

Built for incomes that changes from month to month due to (E.g. dollar exchange rate), so instead of fixed amounts, each category gets a share of whatever comes in.
That way the budget adjusts itself.

**What it does**
- Takes in monthly income

- Splits it into 7 categories: Savings, Food, Faith, Work Tools, Transport, House and Personal Care

- Reads spending records for the month and totals what is spent in each category

- Flags each category as OVER (above budget), CLOSE (80% or more used) or OK

- Treats Savings differently: MET if savings target is reached, SHORT if it doesn't

- Prints a summary with total spent and money left


**How to use it**

Run all cells from the top and enter your monthly income without commas when asked.

**Concepts used**

Variables, input, data types, operators, lists, dictionaries, loops, if/elif/else, functions and f-strings.

```python
# Set up the monthly budget

print("Hello, Let's Build A Personal Budget Tracker For You!\n")

# Income is entered fresh each month because it changes with the exchange rate

income = int(input("What Is Your Monthly Income? (No comma) \n"))

# Each category gets a share of income instead of a fixed amount; the shares add up to 1.00

cat_percentages = {"Savings": 0.20, "Food": 0.20, "Faith": 0.15, "Work Tools": 0.15, "Transport": 0.10, "House": 0.10, "Personal Care": 0.10}

# Loop through each category, work out its budget from income, and store it in budgets (category: amount)

budgets = {}

for category, percentage in cat_percentages.items():
  amount = income * percentage

  budgets[category] = amount
print(budgets)
```

```python
# Spending records for the month, stored as a list of dictionaries (one dictionary per expense)

"""
Every record has the same three keys: category, item and amount

Category names must match the budget categories exactly so the totals line up
"""

expenses = [
    {"category": "Food", "item": "Eggs", "amount": 2650},
    {"category": "Food", "item": "Fish", "amount": 5000},
    {"category": "Faith", "item": "Pledge", "amount": 20000},
    {"category": "Faith", "item": "Tithe", "amount": 5000},
    {"category": "Work Tools", "item": "Laptop Repair", "amount": 30000},
    {"category": "Work Tools", "item": "Charger", "amount": 5000},
    {"category": "Work Tools", "item": "Data", "amount": 3000},
    {"category": "Transport", "item": "Event Transport", "amount": 5000},
    {"category": "House", "item": "Levy", "amount": 8000},
    {"category": "Personal Care", "item": "Hair Care", "amount": 17500},
    {"category": "Personal Care", "item": "Hair Oil", "amount": 3000},
    {"category": "Food", "item": "Rice", "amount": 12000}
]

print(expenses)
```

```python
# Add up spending per category
# Start with an empty dictionary to hold the running totals

totals = {}

# For each expense, pull out its category and amount

for expense in expenses:
  category = expense["category"]
  amount = expense["amount"]

# If the category already has a total, add to it; if it is new, start its total with this amount

  if category in totals:
    totals[category] += amount

  else:
    totals[category] = amount

print(totals)
```

```python
# Decide the status for one category

def get_status(category, spent, budget):

  """ # Savings works the other way: reaching the target is MET, falling below it is SHORT. """

  if category == "Savings":
    if spent >= budget:
      return "MET"
    return "SHORT"

# Spending categories check the most serious case first: OVER, then CLOSE (80% or more used), then OK

  if spent > budget:
    return "OVER"
  if spent >= 0.80 * budget:
    return "CLOSE"
  return "OK"
```

```python
get_status("Savings", 25000, 20000)
```

```python
# Print the budget summary

print("BUDGET SUMMARY\n")

"""
Looping over budgets (not totals) so every category shows, including Savings that has no record

.get(category, 0) gives 0 when a category has no spending, instead of crashing
"""

for category, budget in budgets.items():
  spent = totals.get(category, 0)

  status = get_status(category, spent, budget)

  if status == "OVER":
    print(f"{category}: spent {spent:,.0f} of {budget:,.0f} ({status}, over by {spent - budget:,.0f})")
  else:
    print(f"{category}: spent {spent:,.0f} of {budget:,.0f} ({status})")

total_spent = 0
for amount in totals.values():
  total_spent += amount
money_left = income - total_spent

# Add up all category totals and work out money left from income

print(f"\nIncome: {income:,}")
print(f"\nTotal spent: {total_spent:,}")
print(f"\nMoney left: {money_left:,}")
```
