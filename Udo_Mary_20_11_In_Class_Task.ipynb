The dataset used evaluates leading computer vision researchers whose work shaped modern AI—ranging from deep neural networks (ResNet, VGGNet) and object detection (R-CNN) to 3D geometry and human body modeling (SMPL).

```python
cv_scholars = [
    {"scholar": "Andrew", "levels": [96, 99, 98]},
    {"scholar": "Jitendra", "levels": [92, 90, 97]},
    {"scholar": "Luc", "levels": [90, 100, 92]},
    {"scholar": "Trevor", "levels": [88, 91, 94]},
    {"scholar": "Cordelia", "levels": [85, 91, 92]},
    {"scholar": "Kaiming", "levels": [100, 87, 100]}
]

for sch in cv_scholars:
  print(sch["scholar"], sch["levels"])
```

```python
avgs = {}

for sch in cv_scholars:
  avg = sum(sch["levels"]) / len(sch["levels"])
  avgs[sch["scholar"]] = round(avg)

for scholar in avgs:
  print(scholar, avgs[scholar])
```

```python
top_scholars = []
for scholar in avgs:
  top_scholars.append((avgs[scholar], scholar))

top_scholars.sort(reverse=True)

for avg, scholar in top_scholars[:3]:
  print(scholar, avg)
```

```python
contributors = {"Top": 0, "Middle": 0, "Bottom": 0}

for avg in avgs.values():
  if avg >= 95:
    contributors["Top"] += 1

  elif avg >= 92:
    contributors["Middle"] += 1

  else:
    contributors["Bottom"] += 1

print(contributors)
```

```python
for position in contributors:
  print(f"{position} Contributors Are - {contributors[position]}")
```
