```python
name = input("Elixir Haven Here, What Is Your Name?\n")

scent = input(f"Welcome {name}, Scent A, B, C, or D?\n")
scent = scent.upper()

qty = int(input(f"What Quantity Of {scent} Do You Want?\n"))

if scent == "A":
  price = 6000 * qty

elif scent == "B":
  price = 8000 * qty

elif scent == "C":
  price = 10000 * qty

else:
  price = 12000 * qty

cost = price // qty

print(f"Your Scent Selection {scent} Cost NGN{cost:,.2f}\n")

print(f"NGN{cost:,.2f} x {qty} = NGN{price:,.2f}\n")

if price >= 8000:
  discount = price * 0.05

elif price >= 12000:
  discount = price * 0.08

elif price >= 20000:
  discount = price * 0.15

print(f"Discount Applied Is NGN{discount:,.2f}\n")

total = price - discount

print(f"NGN{price:,.2f} - NGN{discount:,.2f}\n")

print(f"Your Total Purchase Is NGN{total:,.2f}")


```
